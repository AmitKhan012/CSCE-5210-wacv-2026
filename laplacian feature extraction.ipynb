```python
from PIL import Image
import os
import scipy
import numpy as np
import pickle
```

```python
# extracting laplacian features for affected fruit
```

```python
folder_path_affected_fruit = "Dataset/affected"
```

```python
affected_image_names = [f for f in os.listdir(folder_path_affected_fruit)]
```

```python
affected_image_names = sorted(affected_image_names)
```

```python
image_list_affected = []
```

```python
for affected_image_name in affected_image_names:
    affected_image_path = os.path.join(folder_path_affected_fruit, affected_image_name)
    image = Image.open(affected_image_path)
    image_list_affected.append(image.copy())
    image.close()
```

```python
len(image_list_affected)
```

```python
affected_laplacian_zero_dimension = []
affected_laplacian_first_dimension = []
affected_laplacian_second_dimension = []
```

```python
for i in range(len(image_list_affected)):
    affected_laplacian = scipy.ndimage.filters.laplace(image_list_affected[i])

    affected_laplacian_zero = np.matrix(affected_laplacian[:,:,0])
    affected_laplacian_first = np.matrix(affected_laplacian[:,:,1])
    affected_laplacian_second = np.matrix(affected_laplacian[:,:,2])

    affected_laplacian_zero_dimension.append(affected_laplacian_zero.sum())
    affected_laplacian_first_dimension.append(affected_laplacian_first.sum())
    affected_laplacian_second_dimension.append(affected_laplacian_second.sum())
```
