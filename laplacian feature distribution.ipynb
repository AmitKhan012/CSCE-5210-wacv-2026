```python
import pickle
import matplotlib.pyplot as plt
```

```python
filename1 = "affected_laplacian_zero_dimension.pkl"
```

```python
filename2 = "affected_laplacian_first_dimension.pkl"
```

```python
filename3 = "affected_laplacian_second_dimension.pkl"
```

```python
filename4 = "healthy_laplacian_zero_dimension.pkl"
```

```python
filename5 = "healthy_laplacian_first_dimension.pkl"
```

```python
filename6 = "healthy_laplacian_second_dimension.pkl"
```

```python
with open(filename1, "rb") as f:
    affected_laplacian_zero_dimension = pickle.load(f)
```

```python
with open(filename2, "rb") as f:
    affected_laplacian_first_dimension = pickle.load(f)
```

```python
with open(filename3, "rb") as f:
    affected_laplacian_second_dimension = pickle.load(f)
```

```python
with open(filename4, "rb") as f:
    healthy_laplacian_zero_dimension = pickle.load(f)
```

```python
with open(filename5, "rb") as f:
    healthy_laplacian_first_dimension = pickle.load(f)
```

```python
with open(filename6, "rb") as f:
    healthy_laplacian_second_dimension = pickle.load(f)
```

```python
# Plot histograms
plt.hist(affected_laplacian_zero_dimension, bins=10, alpha=0.6, label='affected', color='blue', edgecolor='black')
plt.hist(healthy_laplacian_zero_dimension, bins=10, alpha=0.6, label='healthy', color='orange', edgecolor='black')

# Add labels and legend
plt.xlabel('Feature Value')
plt.ylabel('Frequency')
plt.title('Laplacian feature for Red Color Channel Distribution')
plt.legend()

plt.show()
```
