```python
import cv2
import os
```

```python
folder_path_remove = "Dataset/affected_remove"
```

```python
image_files_remove = [f for f in os.listdir(folder_path_remove) if f.endswith(('.jpg'))]
```

```python
for idx, filename in enumerate(image_files_remove, start=1):
    ext = os.path.splitext(filename)[1]

    # defining new name
    new_name = f'img{idx}{ext}'
    # print(new_name)

    # setting old and new path
    old_path = os.path.join(folder_path_remove, filename)
    new_path = os.path.join(folder_path_remove, new_name)

    os.rename(old_path, new_path)
```
